```python
from syncode import Syncode

grammar = """ start: number operator number

              number: /[0-9]+\.[0-9]+/

              operator: "+" | "-" | "*" | "/"
      """
model_name = "microsoft/phi-2"

example_question1 = "Question: what is 893.234 times 234.23"
example_answer1 = "Extraction:893.234*234.23"

example_question2 = "Question: 12.3 substract 84.0"
example_answer2 = "Extraction:12.3-84.0"

example_question3 = "Question: yoo, how is the sum of 2 and 4"
example_answer3 = "Extraction:2.0+4.0"

example_question4 = "Question: way, you say 749.0 divided by 12.0"
example_answer4 = "Extraction:749.0/12.0"

real_question = ""
real_answer = "Extraction:"

# Load the Syncode augmented model
syn_llm = Syncode(model=model_name, grammar=grammar, parse_output_only=True)
```

Making a function for syncode

```python
def run_syncode(textual_query):
    real_question = f"Question: {textual_query}"
    inp = f"{example_question1}\n{example_answer1}\n{example_question2}\n{example_answer2}\n{example_question3}\n{example_answer3}\n{example_question4}\n{example_answer4}\n{real_question}\n{real_answer}"
    output = syn_llm.infer(inp)
    text = output[0]
    print(text)
    for i in range(1, len(text)):
      if text[i] in "+-*/":
          operator = text[i]
          pos = i
          break
    parts = text.split(operator)
    operands = [float(parts[0]), float(parts[1])]
    if operator == "+":
      answer = operands[0] + operands[1]
    elif operator == "-":
      answer = operands[0] - operands[1]
    elif operator == "*":
      answer = operands[0] * operands[1]
    elif operator == "/":
      answer = operands[0] / operands[1]
    return answer
```

Test

```python
test_question1 = "What is 327. multiplied by 11.0?"
test_question2 = "What is 45.1 plus 23.54?"
test_question3 = "What is 120.4 divided by 8.0?"

test_answer1 = run_syncode(test_question1)
print(test_answer1)

test_answer2 = run_syncode(test_question2)
print(test_answer2)

test_answer3 = run_syncode(test_question3)
print(test_answer3)
```
